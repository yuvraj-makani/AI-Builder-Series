A beginner-friendly Computer Vision project demonstrating image smoothing and edge detection techniques using **OpenCV**, **NumPy**, and **Matplotlib**.


# Import Required Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

# Load Image


```python
img = cv2.imread('./sample_images/gorilla.jpg') 
```

```python


def load_img():
    img = cv2.imread('./sample_images/gorilla.jpg').astype(np.float32) / 255
    return img
```

# Display Original Image

```python
plt.imshow(img)
```

# Gaussian Blur

Gaussian Blur smooths the image by reducing noise and fine details. It is commonly used as a preprocessing step before edge detection to improve the quality of detected edges.

```python
gaussian_img = cv2.GaussianBlur(img, (19,19), 0)
```

```python
plt.title('Gaussian Blur')
plt.axis('off')
plt.imshow(gaussian_img)
```

# Median Blur

Median Blur replaces each pixel with the median value of its neighboring pixels.
Unlike Gaussian Blur, it is highly effective for removing Salt & Pepper Noise while preserving edges.
It is commonly used before edge detection and image segmentation.

```python
median = cv2.medianBlur(img, 25)
plt.title('Median Filter')
plt.axis('off')
plt.imshow(median)
```

# Bilateral Filter

Bilateral Filter smooths the image while preserving sharp edges.
Unlike Gaussian Blur, it considers both the spatial distance and intensity difference between pixels.
It is widely used before edge detection because it removes noise without destroying important boundaries.

```python
bilateral = cv2.bilateralFilter(img, 20,75,75)
plt.title('Bilateral Filter')
plt.axis('off')
plt.imshow(bilateral)
```

# Canny Edge Detection

Canny Edge Detection is a popular edge detection algorithm used to identify object boundaries in an image.

It works in multiple stages:
Noise Reduction (Gaussian Blur),
Gradient Calculation,
Non-Maximum Suppression,
Double Thresholding,
Edge Tracking.


30–100 → A lot of edges.
100–200 → Balanced output. (Most common)
200–300 → Only strong edges remain.

```python
canny_effect = cv2.Canny(img, 100,200)
plt.title('Canny Edge Detection')
plt.axis('off')
plt.imshow(canny_effect ,  cmap='gray')
```

# Save Processed Images

```python
cv2.imwrite('./outputs/Gaussian Blur.jpg', gaussian_img)
cv2.imwrite('./outputs/Canny effect.jpg', canny_effect)
cv2.imwrite('./outputs/Bilateral filter.jpg', bilateral)
cv2.imwrite('./outputs/Median Filter.jpg', median)
```

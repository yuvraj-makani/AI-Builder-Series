# Project: OpenCV-PhotoLab

A beginner-friendly image editing project built using **Python**, **OpenCV**, and **Matplotlib**. This notebook demonstrates fundamental image processing operations such as resizing, flipping, cropping, brightness adjustment, rotation, and saving processed images.


Import Required Libraries

```python
import cv2
import matplotlib.pyplot as plt 
import numpy as np 
```

Load Image

```python
img = cv2.imread('/Users/yuvrajmakani/Desktop/AI-Builder-Series/OpenCV-PhotoLab/sample_images/giraffes.jpg')
```

Display original image

```python
plt.imshow(img)
```

```python
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img)
```

Image information

```python
img.shape
```

Image resizing

```python
resized_img = cv2.resize(img,(700,400))
```

```python
plt.imshow(resized_img)
```

```python
# 0.5 for 50% decrease in size 
w = 0.25   
h = 0.25
ratio_img = cv2.resize(img,(0,0),img,h,w)
```

```python
plt.imshow(ratio_img)
```

Image flipping

```python
# flip image 0 for horizontal, 1 for vertical, -1 for both
flipped_img_h = cv2.flip(img,0)
flipped_img_v = cv2.flip(img,1)
flipped_img_b = cv2.flip(img,-1)
```

```python
plt.imshow(flipped_img_b)
```

Image cropping

```python
crop_img = img[200:1200 , 200:750]
```

```python
plt.imshow(crop_img)
```

```python
crop_img.shape
```

Brightness adjustment

```python
# alpha for contrast and beta for brigtness
bright_img = cv2.convertScaleAbs(img, alpha=1, beta=75)
dark_img = cv2.convertScaleAbs(img, alpha=1, beta=-75)
```

```python
plt.imshow(dark_img)
```

Image rotation

```python
rotate_90 = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
rotate_180 = cv2.rotate(img, cv2.ROTATE_180)
rotate_270 = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)

```

```python
plt.imshow(rotate_270)
```

Save images

```python
cv2.imwrite(
    "outputs/dark_image.jpg",
    cv2.cvtColor(dark_img, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "outputs/bright_image.jpg",
    cv2.cvtColor(bright_img, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "outputs/resize_by_ratio.jpg",
    cv2.cvtColor(ratio_img, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "outputs/resize_by_pixel.jpg",
    cv2.cvtColor(resized_img, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "outputs/cropped_image.jpg",
    cv2.cvtColor(crop_img, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "outputs/horizontal_flipped.jpg",
    cv2.cvtColor(flipped_img_h, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "outputs/vertical_flipped.jpg",
    cv2.cvtColor(flipped_img_v, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "outputs/both_side_flipped.jpg",
    cv2.cvtColor(flipped_img_b, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "outputs/image_rotation_by_90.jpg",
    cv2.cvtColor(rotate_90, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "outputs/image_rotation_by_180.jpg",
    cv2.cvtColor(rotate_180, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "outputs/image_rotation_by_270.jpg",
    cv2.cvtColor(rotate_270, cv2.COLOR_RGB2BGR)
)
```

## Project Overview

Image Explorer Pro is a beginner-friendly Computer Vision project built using Python, OpenCV, NumPy, and Matplotlib.

The project allows users to load an image, analyze its properties, visualize individual color channels, convert it to grayscale, generate a negative image, and save all processed outputs.

This project was developed as part of the AI Builder Series to strengthen the fundamentals of image processing and Computer Vision.

## Import Required Libraries

This section imports all the libraries required for image processing, visualization, and numerical computations.

```python
# IMPORT LIBRARIES

import numpy as np
import matplotlib.pyplot as plt
import cv2
```

## Load Image

In this section, an image is loaded from the local system using OpenCV. The loaded image is stored as a NumPy array for further processing.

```python
img = cv2.imread("/Users/yuvrajmakani/Desktop/CV Projects/Image-Explorer-Pro/sample_images/cat.jpg")
```

## Display Original Image

The loaded image is displayed using Matplotlib after converting it from OpenCV's BGR color format to the RGB format expected by Matplotlib.

```python
plt.imshow(img)
```

```python
#CONVERTING IMAGE FROM BGR TO RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img)
```

```python
# IMAGE TYPE
type(img)
```

```python
#IMAGE SHAPE
print("height:", img.shape[0])
print("width:", img.shape[1])
print("channels:", img.shape[2])
```

```python
print("data type:",img.dtype)
print("total pixels;",img.size)
print("minimum pixel value:", img.min())
print("maximum pixel value:", img.max())
```

```python
#creating duplicate image & displaying only blue channel & saving it as a new image
blue = img.copy()
blue [:,:,0]=0
blue [:,:,1]=0
plt.imshow(blue)
cv2.imwrite("/Users/yuvrajmakani/Desktop/CV Projects/Image-Explorer-Pro/outputs/blue_channel.jpg", blue)
```

```python
#creating duplicate image & displaying only green channel
green = img.copy()
green [:,:,0]=0
green [:,:,2]=0
plt.imshow(green)
cv2.imwrite("/Users/yuvrajmakani/Desktop/CV Projects/Image-Explorer-Pro/outputs/green_channel.jpg", green)
```

```python
#creating duplicate image & displaying only red channel
red = img.copy()
red [:,:,1]=0
red [:,:,2]=0
plt.imshow(red)
cv2.imwrite("/Users/yuvrajmakani/Desktop/CV Projects/Image-Explorer-Pro/outputs/red_channel.jpg", red)
```

```python
gray = img.copy()
plt.imshow(gray, cmap='gray')
cv2.imwrite("/Users/yuvrajmakani/Desktop/CV Projects/Image-Explorer-Pro/outputs/gray_channel.jpg", gray)
```

```python
# negative of the image
negative = 255 - img
```

```python
plt.imshow(negative)
cv2.imwrite("/Users/yuvrajmakani/Desktop/CV Projects/Image-Explorer-Pro/outputs/negative_image.jpg", negative)
```
